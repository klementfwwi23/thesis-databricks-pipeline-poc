```python
declarative_color = "#0F2DB3"
imperative_color = "#0072EF"
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import os

# ---------------------------------------------------------
# Daten: Sekunden -> Minuten umrechnen
# ---------------------------------------------------------
data = {
    "pipeline": ["declarative", "declarative", "imperative", "imperative"],
    "data_scale": ["1x", "2x", "1x", "2x"],
    # Sekunden -> Minuten
    "avg_runtime_minutes": [
        843.2 / 60,
        751.67 / 60,
        1417.6 / 60,
        1487.33 / 60,
    ],
    "avg_cost_usd_per_run": [18.39, 22.84, 28.39, 49.67],
}

kpi_df = pd.DataFrame(data)

change_df = pd.DataFrame({
    "pipeline": ["declarative", "imperative"],
    "runtime_change_pct_per_doubling": [-10.86, 4.92],
    "cost_change_pct_per_doubling": [24.22, 74.97],
})

summary_df = pd.DataFrame({
    "pipeline": ["declarative", "imperative"],
    # Sekunden -> Minuten
    "avg_runtime_1x_minutes": [843.2 / 60, 1417.6 / 60],
    "avg_runtime_2x_minutes": [751.67 / 60, 1487.33 / 60],
    "avg_cost_1x_usd": [18.39, 28.39],
    "avg_cost_2x_usd": [22.84, 49.67],
    "runtime_change_pct_per_doubling": [-10.86, 4.92],
    "cost_change_pct_per_doubling": [24.22, 74.97],
})

os.makedirs("output", exist_ok=True)

# ---------------------------------------------------------
# Farben
# ---------------------------------------------------------
if "declarative_color" not in globals():
    declarative_color = "#1f77b4"
if "imperative_color" not in globals():
    imperative_color = "#ff7f0e"

# ---------------------------------------------------------
# Figure 1: Average end-to-end runtime by pipeline and data scale (Minuten)
# ---------------------------------------------------------
scales = ["1x", "2x"]
x = np.arange(len(scales))
width = 0.35

dec_runtime = [
    kpi_df.loc[(kpi_df["pipeline"] == "declarative") & (kpi_df["data_scale"] == s),
               "avg_runtime_minutes"].iloc[0]
    for s in scales
]
imp_runtime = [
    kpi_df.loc[(kpi_df["pipeline"] == "imperative") & (kpi_df["data_scale"] == s),
               "avg_runtime_minutes"].iloc[0]
    for s in scales
]

fig, ax = plt.subplots(figsize=(6, 4))
ax.bar(x - width/2, dec_runtime, width, label="Declarative", color=declarative_color)
ax.bar(x + width/2, imp_runtime, width, label="Imperative", color=imperative_color)
ax.set_xticks(x)
ax.set_xticklabels(scales)
ax.set_ylabel("Average runtime (min)")
ax.set_title("Average end-to-end runtime by pipeline and data scale")
ax.legend()

for spine in ax.spines.values():
    spine.set_visible(False)

plt.tight_layout()
plt.show()

# ---------------------------------------------------------
# Figure 2: Average cost per run by pipeline and data scale
# ---------------------------------------------------------
dec_cost = [
    kpi_df.loc[(kpi_df["pipeline"] == "declarative") & (kpi_df["data_scale"] == s),
               "avg_cost_usd_per_run"].iloc[0]
    for s in scales
]
imp_cost = [
    kpi_df.loc[(kpi_df["pipeline"] == "imperative") & (kpi_df["data_scale"] == s),
               "avg_cost_usd_per_run"].iloc[0]
    for s in scales
]

fig, ax = plt.subplots(figsize=(6, 4))
ax.bar(x - width/2, dec_cost, width, label="Declarative", color=declarative_color)
ax.bar(x + width/2, imp_cost, width, label="Imperative", color=imperative_color)
ax.set_xticks(x)
ax.set_xticklabels(scales)
ax.set_ylabel("Average cost per run (USD)")
ax.set_title("Average cost per run by pipeline and data scale")
ax.legend()

for spine in ax.spines.values():
    spine.set_visible(False)

plt.tight_layout()
plt.show()

# ---------------------------------------------------------
# Figure 3: Runtime and cost change per data doubling
# (unverändert, da es Prozentwerte sind)
# ---------------------------------------------------------
metrics = ["runtime_change_pct_per_doubling", "cost_change_pct_per_doubling"]
metric_labels = ["Runtime change (%)", "Cost change (%)"]
x = np.arange(len(metrics))
width = 0.35

dec_change = [change_df.loc[change_df["pipeline"] == "declarative", m].iloc[0]
              for m in metrics]
imp_change = [change_df.loc[change_df["pipeline"] == "imperative", m].iloc[0]
              for m in metrics]

fig, ax = plt.subplots(figsize=(6, 4))
ax.bar(x - width/2, dec_change, width, label="Declarative", color=declarative_color)
ax.bar(x + width/2, imp_change, width, label="Imperative", color=imperative_color)
ax.set_xticks(x)
ax.set_xticklabels(metric_labels)
ax.set_ylabel("Change (%)")
ax.set_title("Runtime and cost change per data doubling")
ax.legend()

for spine in ax.spines.values():
    spine.set_visible(False)

plt.tight_layout()
plt.show()

# ---------------------------------------------------------
# Summary-Table
# ---------------------------------------------------------
summary_df
summary_df.to_csv("output/kpi_summary_table.csv", index=False)
```
